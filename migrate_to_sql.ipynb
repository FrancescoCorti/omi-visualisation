```python
import pandas as pd
from sqlalchemy import create_engine
```

SQL ENGINE

```python
engine = create_engine("postgresql://postgres:Francesco.24@localhost:5432/municipalities")
```

ISTAT CODES

```python
df = pd.read_parquet("datasets/mun_istat_codes.parquet")

new_col = [
    "mun_istat",
    "mun_name_norm",
    "prov_name_norm",
    "reg_name_norm"
]

df = df[new_col]

df.to_sql(
    "mun_info",
    engine,
    if_exists = "replace",
    index = False
)
```

PRICE SEMESTER

```python
df = pd.read_parquet("datasets/house_updated_temp.parquet")

map_type = {
    "lowcost housing" : 1,
    "residential housing" : 2,
    "independent houses and villas" : 3
}

df["type"] = df["type"].replace(map_type)

map_cond = {
    "poor" : 1,
    "normal" : 2,
    "excellent" : 3
}

df["condition"] = df["condition"].replace(map_cond)

df = df.rename(columns =  {"mun_istat_updated" : "mun_istat"})
 
new_col = [
    "mun_istat",
    "year",
    "year_semester",
    "semester",
    "zone",
    "type",
    "condition",
    "buy_min",
    "buy_max"
]

df = df[new_col]

df.to_sql(
    "mun_price",
    engine,
    if_exists = "replace",
    index = False
)
```

DWELLINGS

```python
df = pd.read_parquet("datasets/mun_dwelling_temp.parquet")

df = df.rename(columns = {
    "mun_istat_updated" : "mun_istat",
    "dwellings" : "dwelling_num"
    })

df = df[["mun_istat", "dwelling_num"]]

df.to_sql(
    "mun_dwelling",
    engine,
    if_exists = "replace",
    index = False
)
```

TOURISM

```python
df = pd.read_parquet("datasets/mun_tourism_temp.parquet")
```

```python
df = df.drop(columns = ["mun_name", "prov_name", "reg_name", "prov_name_norm", "reg_name_norm", "mun_key"])

df = df.rename(columns = {"mun_istat_updated" : "mun_istat"})

df.insert(0, 'mun_istat', df.pop('mun_istat'))

df.to_sql(
    "mun_tourism",
    engine,
    if_exists = "replace",
    index = False
)
```

POPULATION

```python
df = pd.read_parquet("datasets/mun_dem_temp.parquet")
```

```python
df = df.rename(columns = {"mun_istat_updated": "mun_istat"})

df = df[["mun_istat", "year", "population"]]

df.to_sql(
    "mun_pop",
    engine,
    if_exists = "replace",
    index = False
)
```

INCOME

```python
df = pd.read_parquet("datasets/mun_income_temp.parquet")
```

```python
df = df.rename(columns = {"mun_istat_updated" : "mun_istat"})

df = df[["mun_istat", "year", "median_income"]]

df.to_sql(
    "mun_income",
    engine,
    if_exists = "replace",
    index = False
)
```
