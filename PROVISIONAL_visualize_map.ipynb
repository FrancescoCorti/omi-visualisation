```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import contextily as cx
```

Load data

```python
df = gpd.read_file('datasets/geo_data/mun_perimeters/mun_perimeters.gpkg') 
```

Map - average prices

```python
fig, ax = plt.subplots(1, 1, figsize=(12, 10))

df = df.to_crs(epsg=3857)

df.plot(
    linewidth=0.2,
    edgecolor="black",
    legend=True,
    ax=ax
)

cx.add_basemap(
    ax,
    source=cx.providers.CartoDB.PositronNoLabels
)

ax.set_axis_off()

plt.tight_layout()

plt.show()
```
