```python
import pandas as pd
import geopandas as gpd
import folium
from functions import *
```

Zone map - max. 3 provinces (2 is better)

```python
# Load zone data
gdf_zone = gpd.read_file('datasets/data_maps/zone_estimate.gpkg')
```

```python
gdf_select = gdf_zone[gdf_zone['prov_name'].isin(['Milano', 'Monza e della Brianza'])]
```

```python
m = gdf_select.explore(
    tiles="CartoDB.DarkMatter",
    column="buy_max",
    style_kwds=dict(fillOpacity=0.3, weight=0.5),
    highlight_kwds=dict(fillOpacity=0.3, weight=1, color="white")
)

# Add custom CSS for white legend text
legend_css = """
<style>
.legend {
    color: white !important;
    background-color: rgba(0, 0, 0, 0.7) !important;
}
.legend text {
    fill: white !important;
}
</style>
"""

m.get_root().html.add_child(folium.Element(legend_css))

m.save("maps/map_zone.html")
```

Municipality map - max. 1 region (good with smaller regions; slower with Lombardia)

```python
# Load Municipality data
gdf_mun = gpd.read_file('datasets/data_maps/mun_estimate.gpkg')
gdf_mun["geometry"] = gdf_mun["geometry"].simplify(tolerance=50, preserve_topology=True)
```

```python
gdf_select = gdf_mun[gdf_mun['reg_name'].isin(['Lombardia'])]
```

```python
m = gdf_select.explore(
    tiles="CartoDB.DarkMatter",
    column="buy_max",
    style_kwds=dict(fillOpacity=0.3, weight=0.5),
    highlight_kwds=dict(fillOpacity=0.3, weight=1, color="white")
)

# Add custom CSS for white legend text
legend_css = """
<style>
.legend {
    color: white !important;
    background-color: rgba(0, 0, 0, 0.7) !important;
}
.legend text {
    fill: white !important;
}
</style>
"""

m.get_root().html.add_child(folium.Element(legend_css))

m.save("maps/map_mun.html")
```
