```python
import pandas as pd
import string
import re
from functions import *
```

Load data

```python
df = pd.read_csv('data_origin/istat_codes/Elenco dei codici e delle denominazioni delle unit_ territoriali.csv', sep = ';')
```

Translate and clean

```python
dict = {
    'Comune':'mun_name',
    'Provincia/Uts' : 'prov_name',
    'Codice Comune (alfanumerico)':'mun_istat',
    'Codice catasto' : 'land_code',
    'Codice Provincia (Storico)' : 'prov_istat',
    'Regione' : 'reg_name'
}

df = df.rename(columns = dict)

# Order columns
order = ['mun_istat','mun_name','prov_name','prov_istat','reg_name', 'land_code']

df = df[order]

# Add missing heading 0s to Istat codes
add_zeroes(df, ['mun_istat'], 6)

df['mun_istat'] = df['mun_istat'].astype('str')

# Add missing heading 0s to provincial Istat codes
add_zeroes(df, ['prov_istat'], 3)

df['prov_istat'] = df['prov_istat'].astype('str')
```

```python
# Save dataset with non-normalized names
df.to_csv('datasets/mun_istat_codes_non_normalized.csv', index = False)
```

```python
# Normalize municipalities, province, and region names
df['mun_name'] = df['mun_name'].apply(normalize_name)
df['prov_name'] = df['prov_name'].apply(normalize_name)
df['reg_name'] = df['reg_name'].apply(normalize_name)

# Fix normalized names
df['prov_name'] = df['prov_name'].replace({
    'BARLETTAANDRIATRANI' : 'BARLETTA - ANDRIA - TRANI',
    'BOLZANOBOZEN' : 'BOLZANO - BOZEN',
    'FORLICESENA' : 'FORLI - CESENA',
    'MASSACARRARA' : 'MASSA - CARRARA',
    "VALLE D'AOSTAVALLEE D'AOSTE" : "VALLE D'AOSTA",
    'VERBANOCUSIOOSSOLA' : 'VERBANO - CUSIO - OSSOLA'
})

df['reg_name'] = df['reg_name'].replace({
    'EMILIAROMAGNA' : 'EMILIA ROMAGNA',
    'FRIULIVENEZIA GIULIA' : 'FRIULI VENEZIA GIULIA',
    'TRENTINOALTO ADIGESUDTIROL' : 'TRENTINO ALTO ADIGE - SUDTIROL',
    "VALLE DAOSTAVALLEE DAOSTE" : "VALLE D'AOSTA"
})
```

Save the data

```python
df.to_csv('datasets/mun_istat_codes.csv', index = False, sep = ',')
```
